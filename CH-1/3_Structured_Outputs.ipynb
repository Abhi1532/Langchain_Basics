**Guiding In Prompts**

```python
import os
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI
```

```python
load_dotenv()
```

```python
user_input = input("Enter topic to know fun fact of :")

dynamic_prompt = PromptTemplate.from_template("Write a fun fact about {topic}. Return the result in key value pair with key as 'fact' and value as fun fact")

ready_prompt = dynamic_prompt.invoke({"topic" : user_input})
llm_openai = ChatOpenAI(model="gpt-3.5-turbo",api_key=os.environ.get("OPENAI_API_KEY"),base_url="https://openrouter.ai/api/v1",temperature=0.9)
llm_openai.invoke(ready_prompt)
```

**Using Pydentic Models**

```python
from pydantic import BaseModel,Field
```

```python
class llm_schema(BaseModel):
    setup: str = Field(description="Setup of the joke")
    punch: str = Field(description="Punchline of the joke")
```

```python
obj = llm_schema(**{"setup": "some setup", "punch": "some punch"})
```

```python
llm_openai = ChatOpenAI(model="gpt-3.5-turbo",api_key=os.environ.get("OPENAI_API_KEY"),base_url="https://openrouter.ai/api/v1",temperature=0.9)
```

```python
llm_structured_output = llm_openai.with_structured_output(llm_schema)
```

```python
llm_structured_output.invoke("Tell me a joke")
```

**Using Typed Dicts**

```python
from typing import TypedDict
```

```python
class llm_schema_td(TypedDict):
    setup: str
    punch: str
```

```python
llm_typeddict_schema=llm_schema_td({"setup":"some setup","punch":"some punchline"})
```
