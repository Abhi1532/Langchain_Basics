**Messages**

```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage

load_dotenv()

from langchain_openai import ChatOpenAI

llm_openai = ChatOpenAI(model="gpt-3.5-turbo", api_key=os.environ.get("OPENAI_API_KEY"),base_url="https://openrouter.ai/api/v1",temperature=0.9)

my_messages = [
    SystemMessage(content="You are an assistant who always speaks very politely and formally"),
    HumanMessage(content="How will you present this sentence: Did you know that honey never spoils? Archaeologists have found pots of honey in ancient Egyptian tombs that are over 3000 years old and still perfectly edible! Talk about forever food, am I right?")
]

llm_openai.invoke(my_messages).content
```

**Prompts**

```python
from langchain_core.prompts import PromptTemplate

user_input =  input("Enter a topic for fun fact :")

dynamic_prompt = PromptTemplate.from_template("Write a fun fact about {topic}")
ready_prompt = dynamic_prompt.invoke({"topic" : user_input})

llm_openai.invoke(ready_prompt).content
```

**Tone**

```python
from langchain_core.prompts import ChatPromptTemplate

user_input = input("Enter a topic for fun fact : ")
chatprompt_template = ChatPromptTemplate.from_messages([
    ("system","You are a rude assistant"),
    ("user","Tell me a fun fact about {topic}")
])


ready_prompt = chatprompt_template.invoke({"topic" : user_input})
llm_openai.invoke(ready_prompt).content
```
