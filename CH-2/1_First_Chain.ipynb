**CHAINS in Langchain** 

```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
load_dotenv()

from langchain_openai import ChatOpenAI

llm_openai = ChatOpenAI(model="gpt-3.5-turbo", api_key=os.environ.get("OPENAI_API_KEY"),base_url="https://openrouter.ai/api/v1",temperature=0.9)
```

**First Chain**

```python
# Task - 1 : Prompt
prompt_template = ChatPromptTemplate.from_messages([
    ("system","You are a helpful assistant"),
    ("human","{input}")
])
```

```python
# Task - 2 : LLM
llm_openai = ChatOpenAI(model="gpt-3.5-turbo", api_key=os.environ.get("OPENAI_API_KEY"),base_url="https://openrouter.ai/api/v1",temperature=0.9)
```

```python
# Task - 3 : Parser
str_parser = StrOutputParser()
```

**Manual Invocation**

```python
template = prompt_template.invoke("wHAT IS THE capital of India")

llm_openai.invoke(template)
```

**Chain Invocation**

```python
chain = prompt_template | llm_openai | str_parser
chain.invoke("What is the capital of France")
```
